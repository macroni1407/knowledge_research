```python
def object_aware_occlusion(image, mask, ratio=0.3, mode="center"):

    occluded = image.copy()
    H, W = mask.shape

    ys, xs = np.where(mask > 0)

    if len(xs) == 0:
        return occluded

    mask_area = len(xs)
    target = int(mask_area * ratio)

    # ==============================
    # STRUCTURED OCCLUSION (return early)
    # ==============================
    if mode == "structured":

        stripe_w = max(5, int(np.sqrt(target)))

        for i in range(0, W, stripe_w*2):
            occluded[:, i:i+stripe_w] = 0

        return occluded

    # ==============================
    # CENTER
    # ==============================
    if mode == "center":

        cx = int(xs.mean())
        cy = int(ys.mean())

    # ==============================
    # EDGE
    # ==============================
    elif mode == "edge":

        boundary = []

        for y, x in zip(ys, xs):
            if mask[max(0,y-1):y+2, max(0,x-1):x+2].sum() < 9:
                boundary.append((y,x))

        if len(boundary) > 0:
            cy, cx = random.choice(boundary)
        else:
            idx = np.random.randint(len(xs))
            cx, cy = xs[idx], ys[idx]

    # ==============================
    # RANDOM
    # ==============================
    else:

        idx = np.random.randint(len(xs))
        cx, cy = xs[idx], ys[idx]

    # ==============================
    # RECTANGLE SIZE
    # ==============================

    aspect = np.random.uniform(0.5,2.0)

    rect_h = int(np.sqrt(target/aspect))
    rect_w = int(rect_h*aspect)

    x1 = int(cx - rect_w//2)
    y1 = int(cy - rect_h//2)

    x2 = min(W, x1 + rect_w)
    y2 = min(H, y1 + rect_h)

    x1 = max(0,x1)
    y1 = max(0,y1)

    occluded[y1:y2, x1:x2] = 0

    return occluded
```

```python
def object_aware_cutmix(image1, mask1, image2, mask2, ratio=0.3):

    mixed_image = image1.copy()
    mixed_mask = mask1.copy()

    H, W = mask1.shape

    ys, xs = np.where(mask1 > 0)

    if len(xs) == 0:
        return mixed_image, mixed_mask

    mask_area = len(xs)
    target = int(mask_area * ratio)

    # ==============================
    # CENTER (object center)
    # ==============================

    cx = int(xs.mean())
    cy = int(ys.mean())

    # ==============================
    # RECTANGLE SIZE
    # ==============================

    aspect = np.random.uniform(0.5, 2.0)

    rect_h = int(np.sqrt(target / aspect))
    rect_w = int(rect_h * aspect)

    x1 = int(cx - rect_w // 2)
    y1 = int(cy - rect_h // 2)

    x2 = min(W, x1 + rect_w)
    y2 = min(H, y1 + rect_h)

    x1 = max(0, x1)
    y1 = max(0, y1)

    # ==============================
    # CUTMIX
    # ==============================

    mixed_image[y1:y2, x1:x2] = image2[y1:y2, x1:x2]
    mixed_mask[y1:y2, x1:x2] = mask2[y1:y2, x1:x2]

    return mixed_image, mixed_mask
```

```python
def coco_annotation_to_mask(ann, height, width):

    segmentation = ann["segmentation"]

    if isinstance(segmentation, list):
        rles = maskUtils.frPyObjects(segmentation, height, width)
        rle = maskUtils.merge(rles)
    elif isinstance(segmentation["counts"], list):
        rle = maskUtils.frPyObjects(segmentation, height, width)
    else:
        rle = segmentation

    mask = maskUtils.decode(rle)
    return mask
```

```python
class DAMRunner:
    def __init__(self):
        print("Loading DAM...")

        self.device = torch.device(DEVICE)

        model = AutoModel.from_pretrained(
            "nvidia/DAM-3B-Self-Contained",
            trust_remote_code=True,
            torch_dtype=torch.float16
        ).to(self.device)

        self.dam = model.init_dam(
            conv_mode="v1",
            prompt_mode="full+focal_crop"
        )

        print("DAM loaded.")

    @torch.no_grad()
    def describe(self, image_pil, mask_np):

        mask_np = (mask_np > 0).astype(np.uint8)
        mask_pil = Image.fromarray(mask_np * 255)

        prompt = "<image>\nDescribe the masked region in detail."

        output = self.dam.get_description(
            [image_pil],
            [mask_pil],
            prompt,
            streaming=False,
            temperature=0.2,
            top_p=0.5,
            max_new_tokens=512
        )

        return output.strip()
```

```python
def format_sam_box_from_coco(bbox):
    x, y, w, h = bbox
    return [[[float(x), float(y), float(x+w), float(y+h)]]]
```

```python
class SAMRunner:
    def __init__(self, model_name="facebook/sam-vit-base"):
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

        self.model = SamModel.from_pretrained(model_name).to(self.device)
        self.processor = SamProcessor.from_pretrained(model_name)

    @torch.no_grad()
    def predict(
        self,
        image,
        input_points=None,
        input_labels=None,
        input_boxes=None,
        multimask=True
    ):

        inputs = self.processor(
            image,
            input_points=input_points,
            input_boxes=input_boxes,
            input_labels=input_labels,
            return_tensors="pt"
        ).to(self.device)

        outputs = self.model(**inputs, multimask_output=multimask)

        masks = self.processor.image_processor.post_process_masks(
            outputs.pred_masks,
            inputs["original_sizes"],
            inputs["reshaped_input_sizes"]
        )[0].squeeze(0)

        if multimask:
            scores = outputs.iou_scores[0]
            best_mask = masks[torch.argmax(scores)]
        else:
            best_mask = masks[0]

        return best_mask.cpu().numpy().astype(np.uint8)
```

```python
def run_occ(ratio=0.3, occ_mode="cutout", OUTPUT_FILE=None):

    coco = COCO(ANNOTATION_FILE)
    dam = DAMRunner()
    sam = SAMRunner()

    ann_ids = coco.getAnnIds()
    # ann_ids = ann_ids[:1000]   # limit to 1k

    results = []

    # Resume support
    if os.path.exists(OUTPUT_FILE):
        with open(OUTPUT_FILE, "r") as f:
            results = json.load(f)
        done_ids = set(r["annotation_id"] for r in results)
    else:
        done_ids = set()

    for ann_id in tqdm(ann_ids):

        if ann_id in done_ids:
            continue

        ann = coco.loadAnns(ann_id)[0]
        image_id = ann["image_id"]
        img_info = coco.loadImgs(image_id)[0]

        image_path = os.path.join(IMAGE_ROOT, img_info["file_name"])

        if not os.path.exists(image_path):
            continue

        image_pil = Image.open(image_path).convert("RGB")
        width, height = image_pil.size

        mask = coco_annotation_to_mask(ann, height, width)

        image_np = np.array(image_pil)
        if occ_mode == "cutout":
            image_occ_np = object_aware_occlusion(image_np, mask, ratio=ratio, mode="center")
        else:
            img2_id = random.choice(coco.getImgIds())
            img2_info = coco.loadImgs(img2_id)[0]

            img2_path = os.path.join(IMAGE_ROOT, img2_info["file_name"])
            image2 = np.array(Image.open(img2_path).convert("RGB"))

            ann2 = coco.loadAnns(coco.getAnnIds(imgIds=img2_id))[0]
            mask2 = coco.annToMask(ann2)

            H, W = mask.shape

            image2 = cv2.resize(image2, (W, H))
            mask2 = cv2.resize(mask2, (W, H), interpolation=cv2.INTER_NEAREST)

            image_occ_np, mixed_mask_np = object_aware_cutmix(image_np, mask, image2, mask2, ratio=ratio)
        # image_occ_np = occlusion_block_by_mask_ratio(
        #             image_np, mask, ann["bbox"], ratio=0.3
        # )
        image_occ_pil = Image.fromarray(image_occ_np)

        input_boxes = format_sam_box_from_coco(ann["bbox"])

        pred_mask = sam.predict(
            image_occ_pil,
            input_points=None,
            input_boxes=input_boxes,
            input_labels=None,
            multimask=False
        )

        try:
            description = dam.describe(image_occ_pil, pred_mask)
        except Exception as e:
            print("Error:", e)
            description = ""

        results.append({
            f"{ann_id}": description
        })

        # results.append({
        #     "image_id": image_id,
        #     "annotation_id": ann_id,
        #     "prediction": description
        # })

        # Save every step (resume safe)
        with open(OUTPUT_FILE, "w") as f:
            json.dump(results, f, indent=2)

        torch.cuda.empty_cache()

    print("Done DAM occ caption.")
    print("Saved to:", OUTPUT_FILE)
```

```python
ROI_PROMPT = """
Describe this image region in detail.

Include:
- object appearance (color, shape, material)
- attributes and state
- any visible text
- surrounding context if helpful

Be precise and factual.
"""
```

```python
from transformers import AutoProcessor, AutoModelForImageTextToText
import torch

# model_name = "Qwen/Qwen2.5-VL-7B-Instruct"
model_name = "meta-llama/Llama-3.2-11B-Vision-Instruct"

processor = AutoProcessor.from_pretrained(
    model_name,
    # max_pixels=768*768         # only for qwen
)

model = AutoModelForImageTextToText.from_pretrained(
    model_name,
    dtype=torch.float16,
    device_map="auto"
)
```

```python
def caption_LLM(model, OUTPUT_FILE):

    coco = COCO(ANNOTATION_FILE)

    ann_ids = coco.getAnnIds()
    # ann_ids = ann_ids[:1000]   # limit to 1k

    results = []

    # Resume support
    if os.path.exists(OUTPUT_FILE):
        with open(OUTPUT_FILE, "r") as f:
            results = json.load(f)
        done_ids = set(r["annotation_id"] for r in results)
    else:
        done_ids = set()

    for ann_id in tqdm(ann_ids):

        if ann_id in done_ids:
            continue

        ann = coco.loadAnns(ann_id)[0]
        image_id = ann["image_id"]
        img_info = coco.loadImgs(image_id)[0]

        image_path = os.path.join(IMAGE_ROOT, img_info["file_name"])

        if not os.path.exists(image_path):
            continue

        img_pil = Image.open(image_path).convert("RGB")
        x, y, w, h = ann["bbox"]
        img_crop_pil = img_pil.crop((x, y, x+w, y+h))

        messages = [
            {
                "role": "user",
                "content": [
                    {"type": "image"},
                    {"type": "text", "text": ROI_PROMPT},
                ],
            }
        ]

        text = processor.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        inputs = processor(
            text=[text],
            images=[img_crop_pil],
            return_tensors="pt"
        ).to(model.device)

        outputs = model.generate(
            **inputs,
            max_new_tokens=200
        )

        generated_tokens = outputs[0][inputs.input_ids.shape[-1]:]

        caption = processor.tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        )

        results.append({
            f"{ann_id}": caption
        })

        # Save every step (resume safe)
        with open(OUTPUT_FILE, "w") as f:
            json.dump(results, f, indent=2)

    print("Done captioning.")
    print("Saved to:", OUTPUT_FILE)
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# model_name = "Qwen/Qwen2.5-7B-Instruct"
model_name = "meta-llama/Meta-Llama-3-8B-Instruct"
# model_name = "mistralai/Mistral-7B-Instruct-v0.3"
# model_name = "google/gemma-2-9b-it"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype=torch.float16,
    device_map="auto"
)

#pip install bitsandbytes accelerate
# model = AutoModelForCausalLM.from_pretrained(
#     "meta-llama/Meta-Llama-3.1-8B-Instruct",
#     device_map="auto",
#     load_in_4bit=True,
# )
```

```python
import re
import json
import inflect
from tqdm import tqdm
```

**Prompt template**

```python
# =====================
# PROMPT TEMPLATE
# =====================

PROMPT_TEMPLATE = """Answer the multiple-choice question based on the text description of an object in an image. You need to follow these rules:

Rules:
1. Do not output any reasoning. Do not perform correction. Please output exactly one answer from the choices for each question. Do not repeat the question.
2. There is no need for exact matching. Please choose the closest option based on the description.

Answer with only the letter.

Description:
{caption}

Question:
{question}

Choices:
{choices}
"""
```

**LLM query**

```python
# =====================
# LLM QUERY
# =====================

# def query_llm(prompt):
#     response = client.chat.completions.create(
#         model=MODEL_NAME,
#         messages=[{"role": "user", "content": prompt}],
#         temperature=0.0,
#         max_tokens=10,
#     )
#     return response.choices[0].message.content.strip()

def query_llm(model, prompt):

    messages = [{"role": "user", "content": prompt}]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(text, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=2,
        do_sample=False,
        temperature=None,
        top_p=None,
        top_k=None,
        pad_token_id=tokenizer.eos_token_id
    )

    # chỉ lấy phần generate
    generated_tokens = outputs[0][inputs.input_ids.shape[-1]:]

    output_text = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    return output_text


def parse_choice(text):

    text = text.upper()

    patterns = [
        r'\b([A-E])\b',
        r'ANSWER[: ]*([A-E])',
        r'OPTION[: ]*([A-E])'
    ]

    for p in patterns:
        match = re.search(p, text)
        if match:
            return match.group(1)

    return None
```

**Evaluation**

```python
def run_eval_data(model, captions, file_save):

    p = inflect.engine()

    scores_pos = []
    scores_neg = []
    results = {}

    # for key in tqdm(list(qa_data.keys())[:1000]):   # chỉ chạy 1k subset
    for key in tqdm(list(qa_data.keys())):
        key = str(key)

        if key not in captions:
            continue

        caption = captions[key]
        class_name = class_names[key]

        # -------- Recognition Question --------
        if p.singular_noun(class_name):
            recognition_q = f"Is it likely that the objects in the description are {class_name}?"
        else:
            recognition_q = f"Is it likely that the object in the description is {p.a(class_name)}?"

        recognition_dict = {
            "question": recognition_q,
            "choices": [("Yes", "correct"), ("No", "incorrect")],
            "type": "recognition"
        }

        question_dicts = [recognition_dict] + qa_data[key]

        pos_scores = []
        neg_scores = []
        recognition_correct = None

        for q in question_dicts:

            choice_text = ""
            for idx, (choice, _) in enumerate(q["choices"]):
                letter = "ABCDE"[idx]
                choice_text += f"{letter}. {choice}\n"

            prompt = PROMPT_TEMPLATE.format(
                caption=caption,
                question=q["question"],
                choices=choice_text
            )

            try:
                pred = query_llm(model, prompt)
                # print("RAW:", pred)
            except Exception as e:
                print("LLM error:", e)
                continue

            # print("RAW OUTPUT:", pred)
            pred_letter = parse_choice(pred)
            # print("pred_letter", pred_letter)

            if pred_letter is None:
                score = 0
            else:
                idx = "ABCDE".index(pred_letter)
                score = q["choices"][idx][1]

            # -------- Scoring Logic --------
            if q["type"] == "recognition":
                recognition_correct = (score == "correct")

            elif q["type"] == "positive":
                if recognition_correct is False:
                    score = min(0, score)
                pos_scores.append(score)

            elif q["type"] == "negative":
                if recognition_correct is False:
                    score = min(0, score)
                neg_scores.append(score)

        if len(pos_scores) == 0:
            continue

        score_pos = sum(pos_scores) / len(pos_scores)
        score_neg = sum(neg_scores) / len(neg_scores) if neg_scores else 0

        scores_pos.append(score_pos)
        scores_neg.append(score_neg)

        results[key] = {
            "score_pos": score_pos,
            "score_neg": score_neg,
            "recognition": recognition_correct
        }

    # =====================
    # FINAL RESULTS
    # =====================

    avg_pos = sum(scores_pos) / len(scores_pos)
    avg_neg = sum(scores_neg) / len(scores_neg)

    print("Average Positive Score:", avg_pos)
    print("Average Negative Score:", avg_neg)
    print("Final Score:", (avg_pos + avg_neg) / 2)

    # Save
    with open(file_save, "w") as f:
        json.dump(results, f, indent=4)

    print(f"Saved to {file_save}")
```
